```python
import ipywidgets as widgets
from IPython.display import display, clear_output

titulo = widgets.HTML("<h2>CALCULADORA DE COMPRA</h2>")
txt_producto = widgets.Text(description="Producto:", placeholder="Nombre")
txt_precio = widgets.FloatText(description="Precio:", value=0.0)
txt_cantidad = widgets.IntText(description="Cantidad:", value=1)
btn_calcular = widgets.Button(description="Calcular", button_style="success")
salida = widgets.Output()

def calcular_compra(b):
    with salida:
        clear_output(wait=True)
        producto = txt_producto.value
        precio = txt_precio.value
        cantidad = txt_cantidad.value

        total = precio * cantidad
        print(f"Producto: {producto} | Total: {total}")

btn_calcular.on_click(calcular_compra)
display(titulo, txt_producto, txt_precio, txt_cantidad, btn_calcular, salida)
```
